```python
# Import required libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo

print("Libraries imported successfully!")
```

```python
# Download Student Performance dataset from UCI

student_performance = fetch_ucirepo(id=320)

# Separate features and target

X = student_performance.data.features.copy()
y = student_performance.data.targets.copy()

print("Dataset downloaded successfully!")
```

```python
# 1. Check dataset dimensions
print("Dataset dimensions:", X.shape)

# 2. Display all feature names
print("\nFeature columns:")
print(X.columns.tolist())

# 3. Check data types and missing values
print("\nDataset information:")
X.info()

# 4. Check missing values
print("\nMissing values:")
print(X.isnull().sum())

# 5. Check duplicate records
print("\nDuplicate rows:", X.duplicated().sum())

# 6. Check target information
print("\nTarget columns:")
print(y.columns.tolist())

print("\nTarget statistics:")
print(y.describe())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(y.iloc[:, 0], bins=15, kde=True)

plt.title("Distribution of Final Student Grades")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Check target and feature columns

print("Feature columns:")
print(X.columns.tolist())

print("\nTarget columns:")
print(y.columns.tolist())

print("\nTarget summary:")
display(y.describe())
```

```python
# Select the final grade as our prediction target

X = X.copy()
y = student_performance.data.targets["G3"].copy()

print("Input features:", X.shape)
print("Target:", y.shape)

print("\nCategorical columns:")
print(X.select_dtypes(include="object").columns.tolist())

print("\nNumerical columns:")
print(X.select_dtypes(include=np.number).columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

# Identify numerical and categorical columns
categorical_features = X.select_dtypes(
    include=["object", "str", "category"]
).columns.tolist()

numerical_features = X.select_dtypes(
    include=np.number
).columns.tolist()

# Numerical preprocessing
numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing pipelines
preprocessor = ColumnTransformer([
    ("numerical", numerical_pipeline, numerical_features),
    ("categorical", categorical_pipeline, categorical_features)
])

print("Preprocessing pipeline created successfully!")
```

```python
from sklearn.linear_model import LinearRegression

# Create complete ML pipeline
model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])

# Train the model
model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Make predictions
y_pred = model.predict(X_test)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MODEL EVALUATION")
print("-------------------------")
print("MAE:", round(mae, 3))
print("RMSE:", round(rmse, 3))
print("R2 Score:", round(r2, 3))
```

```python
from sklearn.ensemble import RandomForestRegressor

# Create Random Forest pipeline
rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=300,
        random_state=42,
        min_samples_leaf=2
    ))
])

# Train model
rf_model.fit(X_train, y_train)

print("Random Forest trained successfully!")
```

```python
# Random Forest predictions
rf_pred = rf_model.predict(X_test)

# Calculate Random Forest metrics
rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2 = r2_score(y_test, rf_pred)

# Compare models
comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest"],
    "MAE": [mae, rf_mae],
    "RMSE": [rmse, rf_rmse],
    "R2 Score": [r2, rf_r2]
})

display(comparison.round(3))
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(y_test, rf_pred, alpha=0.7)

plt.plot(
    [0, 20],
    [0, 20],
    linestyle="--"
)

plt.xlabel("Actual Final Grades")
plt.ylabel("Predicted Final Grades")
plt.title("Random Forest: Actual vs Predicted Grades")

plt.xlim(0, 20)
plt.ylim(0, 20)
plt.grid(alpha=0.3)

plt.show()
```
